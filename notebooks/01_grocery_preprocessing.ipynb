Since we need monthly average price, we could remove duplicates.

```python
# first reamin only needed columns - "Month", "City", "Continent", Breakfast_Basket_USD"
data_grocery = data_grocery[["Month", "City", "Continent", "Breakfast_Basket_USD"]]
data_grocery
```

```python
# the number of duplicates
data_grocery.duplicated().sum()
```

```python
# remove duplicates
data_grocery.drop_duplicates()
```

For the study we need monthly average price of breakfast basket in each continent.

```python
# group by Month and Continent, get the mean value of 'Breakfast_Basket_USD' column and reset index to have Month column
grocery_group_by_continent = data_grocery.groupby(by=["Month", "Continent"])[["Breakfast_Basket_USD"]].mean().reset_index()
grocery_group_by_continent
```

For comparing the prices of oil and breakfast basket by month, we need a dataset that has index: "Month", columns: "Each continent" and "Oil" and values: the prices. 

```python
# set "Month" column as index in the data set.
grocery_group_by_continent.set_index("Month")
grocery_group_by_continent
```

```python
# reshape the data, based on column values
grocery_group_by_continent = grocery_group_by_continent.pivot(index="Month", columns="Continent", values="Breakfast_Basket_USD")
grocery_group_by_continent
```

Store the data set for further usage.
